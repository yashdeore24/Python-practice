# python sets

```python
# A set is a collection of unique elements (no duplicates allowed )
# A set is unordered, so elements do not have a fixed position.(sets are unordered becasuse they use hashing, not indexing.)
# A set is mutable, meaning you can add or remove elements.
# A set does not support indexing or slicing.
# A set can store only immutable (hashable) elements like numbers, strings, tuples.
# A Listm dictionariesm or sets cannot be elements inside a set.
# an empty set is created using set(), not{}
# sets are useful for fast membership chaking
```

```python
# Membership operater
# the logic work on set,list  and tuple
Set = {1,2,3,4,5,6,}

print(2 in Set) # the logic work on set

List = [1,2,3,4,5,6,]
print(2 in List)

Tuple = (1,2,3,4,5,6,)
print(2 in Tuple)
```
