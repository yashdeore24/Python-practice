```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.linspace(0,10,100)
Y = np.sin(X)

plt.plot(X,Y,
         color = 'Red',
         marker = 'o',
         markersize = 6,
         markeredgecolor = 'Blue',
         markerfacecolor = 'Yellow',
         
        )

plt.xlabel("X Axis Data")  # Sets the label for the X-axi
plt.ylabel("Sin X")
plt.title("Sine Wave")
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 6))

# Subplot 1
axes[0, 0].plot(x, np.sin(x))
plt.sca(axes[0, 0])
plt.title("Sine")

# Subplot 2
axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title("Cosine")

# Subplot 3
axes[1, 0].plot(x, x**2)
axes[1, 0].set_title("Square")

# Subplot 4
axes[1, 1].plot(x, x)
axes[1, 1].set_title("Straight Line")

# Title for the entire figure
fig.suptitle("My 2 × 2 Subplots")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)
fig,axis = plt.subplots(2, 2,
               figsize = (12,8),
                facecolor = "lightgreen"
               )
plt.suptitle("Four Mathematical Plots",
            fontsize = 16,
            fontweight = "bold")

# subplot 1 --> sine:
axis[0,0].plot(x,np.sin(x),
               color = "blue",
               linewidth = 2,
              )
axis[0,0].set_title("Sine")
axis[0,0].set_xlabel("X")
axis[0,0].set_ylabel("Sim(x)")
axis[0,0].grid(True,alpha = 0.5,
              linestyle = ":")

# subplot 2 --> cosine
axis[0,1].plot(x,np.cos(x),
              color="red",
              linewidth = 2)
axis[0,1].set_title("Cosine")

# subplot 3 -- Bar:

axis[1,0].bar([1, 2, 3, 4],
             [10, 20, 15, 25],
             color = "green",
             )
axis[1,0].set_title("Bar chart")

# subplot 4 -- scatter

axis[1,1].scatter(x[:30],
                  np.sin(x[:30]),
                  color = "Orange",
                  s = 30)


axis[1,1].set_title("Scatter Plot")

plt.savefig("four_plots.png",
           dpi = 100,
           bbox_inches = "tight")
plt.show()
```
