```python
Total_Vowels = 0
Total_Consonants = 0
Total_Digits = 0
Total_Space = 0
T_S_C = 0 
String = input()
String = String.lower()
for i in String:
    if i == 'a' or i == 'i' or i == 'e' or i == 'o' or i == 'u':
        Total_Vowels+=1
    elif i.isdigit():
        Total_Digits+=1
    elif i == " ":
        Total_Space+=1
    elif i.isalpha():
        Total_Consonants+=1
    else:
        T_S_C+=1


print("Vowels: ",Total_Vowels)
print("Consonants: ",Total_Consonants)
print("Digits: ",Total_Digits)
print("Spaces: ",Total_Space)
if Total_Vowels > Total_Consonants:
    print("More vowels")
else:
    print("More consonants")
```

## 3.Task

```python
larger = 0
smallest = 0
Sum = 0
List = []
# input the 10 numbers from user
for i in range(1,11):
    List.append(int(input()))
    
# find largest using loop   
larger = List[0]
for i in range(10):
    if List[i] > larger:
        larger = List[i] 
    continue

# find smallest using loop
smallest = List[0]
for i in range(10):
    if List[i] < smallest:
        smallest = List[i] 
    continue

# calculate sum of all elements in list 
for i in List:
    Sum+=i

Average = Sum / 10

# list of even and odd number

Elist = []
Odlist = []

for i in List:
    if i % 2 == 0:
        Elist.append(i)
    
    else:
        Odlist.append(i)

print("Original List:",List)
print("\t Larger number:",larger)

print("\t Smallest number:",smallest)
print("\t Sum of the list elements:",Sum)
print("Even list:",Elist)
print("Odd list:",Odlist)

if Average > 50:
    print("Average is Hight")
else:
    print("Average is low")

```

```python

sentence = input("Enter any sentance:")
words= sentence.split()

print("Total number of words: ",len(words))

# find the longest word from the list of strings
larger = words[0]
for i in words:
    if len(i) > len(larger):
        larger = i
print("The larges word is:",larger)

# find the smallest word form the list of strings
smaller = words[0]
for i in words:
    if len(i) < len(smaller):
        smaller = i
print("The smaller word is:",smaller)

# find the words that start with a vowel
Start_vowel = 0
for i in words:
    if i[0] == 'a' or i[0] == 'i' or i[0]=='e' or i[0]=='o' or i[0] == 'u':
        Start_vowel+=1
    continue
print("Number which start with vowel:",Start_vowel)

# find the words that end with a vowel
Start_Evowel = 0
for i in words:
    if i[-1] == 'a' or i[-1] == 'i' or i[-1]=='e' or i[-1]=='o' or i[-1] == 'u':
        Start_Evowel+=1
    continue
print("Number which end with vowel:",Start_Evowel)


even = []
odd = []
for i in words:
    if len(i) % 2 == 0:
        even.append(i)
    else:
        odd.append(i)


if len(even) > len(odd):
    print("More even-length words")
else:
    print("More odd-length words")
    
        
```
