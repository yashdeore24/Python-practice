# 1.Task

```python
# Take a sentence, split it into words, and remove the first word. Join the remaining words.

string = "hi my name is yash"

string = string.split(" ")
print(string)

string.pop(0)
print(string)

string = "".join(string)
print(string)
```

# 2.Task

```python
# Take a sentence, split it into words, and remove the last word. Join the remaining words.

string = "hi my name is yash"

string = string.split(" ")
print(string)

string.pop(-1)
print(string)

string = "".join(string)
print(string)
```

# 3.Task

```python
# Take a sentence and swap the first and last words.

string = "hi my name is yash"

string = string.split(" ")

temp = string[0]
string[0] = string[-1]
string[-1] = temp

print(string)
```

# 4. Task

```python
# Take a sentence and reverse the order of words using split(), slicing, and join()

string = "Hi my name is yash"

string = string.split(" ")
print(string)

string = string[::-1]
print(string)

string = " ".join(string)
print(string)
```

# 5.task

```python
"""Take a string:

apple,,,banana,,mango

Replace every comma with a space, split it, and join the words using a single comma."""

string = "apple,,,banana,,mango"

string = string.replace(","," ")
print(string)

string = string.split()
print(string)



string = ",".join(string)
print(string)
```
