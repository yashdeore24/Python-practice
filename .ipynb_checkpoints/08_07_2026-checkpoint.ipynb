```python
# removre(): remove() method removes the first occurrence of the elements with the specified 
# syntax : List.remove(elements)
fruits = ["Apple", "Mango", "Apple", "Graps","Banana"]
fruits.remove("Apple")
print(fruits)
# At a time you remove singlie element
```

```python
# pop() : the pop() method removes the elementat the specified position 
# syntax: list.pop(index_pos)

fruits =  ["Apple", "Mango", "Apple", "Graps","Banana"]
fruits.pop(4)
print(fruits)
fruits.insert(4,"Banana")
print(fruits)
fruits.pop()# remove last index
print(fruits)
# fruits.pop(11) # error occours because index is out of range
fruits.insert(4,"Banana")

```

```python
# clear() method : remove all elements in the list
fruits =  ["Apple", "Mango", "Apple", "Graps","Banana"]
fruits.clear()
fruits
```

```python
# index() : display the value position


fruits =  ["Apple", "Mango", "Apple", "Graps","Banana"]

print(fruits.index("Banana"))
fruits
# we can't find the multiple elemets index using index method at a time 
# we find gust one elemet at a time using index method
```

```python
# sort() : order the list elements 

# syntax = list.sort(reverse = True/False)
# same type of values i want in list for short using sort functin 
Age = [20,34,54,12,54,34,67]

Age.sort() # by default it is short asssending form
print(Age)

Age.sort(reverse = True) # we use reverse = True for Decending form
print(Age)
```

```python
fruits =  ["Apple", "Mango", "Apple", "Graps","Banana"]

fruits.sort()
print(fruits)

fruits.sort(reverse = True)
print(fruits)
```
