## Task 1: Integer to Float



```python
# Take an integer from the user and convert it to a float.
# Print the value and its data type.
num = int(input("Enter the number:"))

num = float(num)

print(f"Number is converted into 3 to {num } and its type is {type(num)}")
```

## Task 2: Float to Integer





```python
#Take a float from the user.
#Convert it to an integer.
#Print both the original and converted values

num = float(input("Enter the number:"))

num = int(num)

print(f"Number is converted into 3.3 to {num} and its type is {type(num)}")
```

## Task 7: Calculate Average



```python
"""Take marks of 3 subjects using input().
Convert each input to an integer.
Calculate and print the average."""

sub1 = input()
sub2 = input()
sub3 = input()

sub1 = int(sub1)
sub2 = int(sub2)
sub3 = int(sub3)

total = sub1 + sub2 + sub3 
average = total /3

print("Average is :", average)
```

## Task 8: Shopping Bill



```python
"""Take:

Product price
Quantity
Convert them to the correct data types.
Print the total bill."""

print("Poduct Names :- mung daal , masur daal , apple , mango ")
product_Name = str(input("Enter the product Name:"))
Quantity = float(input("Enter the product Quatity:"))

print("Product Name"+"          "+"price"+"          "+"Quantity"+"          "+"Amount")
print('-'*80)
if product_Name == "mung daal":
    print(f"{product_Name}                 100              {Quantity}          {Quantity*100}")
elif product_Name == "masur daal":
    print(f"{product_Name}                 92               {Quantity}           {Quantity*92}")
elif product_Name == "apple":
    print(f"{product_Name}                 200              {Quantity}           {Quantity*200}")
elif product_Name == "mango":
    print(f"{product_Name}                 80               {Quantity}            {Quantity* 80}")



                
```
