```python
import tkinter as tk
from tkinter import messagebox
import random

# --------------------------
# Sample user data
# --------------------------
username = "admin"
password = "1234"

generated_otp = ""

# --------------------------
# Login Function
# --------------------------
def login():
    global username, password

    user = user_entry.get()
    pwd = pass_entry.get()

    if user == username and pwd == password:
        messagebox.showinfo("Login", "Login Successful")
    else:
        messagebox.showerror("Login", "Invalid Username or Password")

# --------------------------
# Send OTP
# --------------------------
def send_otp():
    global generated_otp

    generated_otp = str(random.randint(100000, 999999))

    # Replace this print statement with SMS API
    print("OTP Sent:", generated_otp)

    messagebox.showinfo(
        "OTP",
        "OTP has been sent to your registered phone number.\n(Check console for demo)"
    )

# --------------------------
# Verify OTP and Change Password
# --------------------------
def update_password():
    global password

    entered_otp = otp_entry.get()

    if entered_otp == generated_otp:

        new_pass = new_pass_entry.get()

        if new_pass == "":
            messagebox.showerror("Error", "Enter New Password")
            return

        password = new_pass
        messagebox.showinfo("Success", "Password Updated Successfully")

    else:
        messagebox.showerror("Error", "Incorrect OTP")

# --------------------------
# GUI
# --------------------------
root = tk.Tk()
root.title("Login Page")
root.geometry("350x350")

tk.Label(root, text="Username").pack()
user_entry = tk.Entry(root)
user_entry.pack()

tk.Label(root, text="Password").pack()
pass_entry = tk.Entry(root, show="*")
pass_entry.pack()

tk.Button(root, text="Login", command=login).pack(pady=10)

tk.Label(root, text="Forgot Password?").pack()

tk.Button(root, text="Send OTP", command=send_otp).pack()

tk.Label(root, text="Enter OTP").pack()
otp_entry = tk.Entry(root)
otp_entry.pack()

tk.Label(root, text="New Password").pack()
new_pass_entry = tk.Entry(root, show="*")
new_pass_entry.pack()

tk.Button(root, text="Update Password", command=update_password).pack(pady=10)

root.mainloop()
```
