```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
name= ["prashant","yash","kapil","Rohit"]
explore=[1,0,0,0]

plt.pie(x,startangle = 100,
        labels = name,
        explode = explore,
        shadow =True,
        labeldistance =0.100,
        wedgeprops={"edgecolor": "black", "linewidth": 5})
plt.show()
```

```python

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 300, 150, 50]
plt.pie(sales,labels = products,
       autopct="%1.1f%%",
    )

plt.title("Product Sales Distribution")
plt.show()
```
