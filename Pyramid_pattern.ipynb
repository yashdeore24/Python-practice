```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Sample dataset
df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [10, 20, 30, 40, 50, 60, 70, 80],
    "Previous_Marks": [30, 35, 38, 50, 55, 60, 72, 80],
    "Sleep_Hours": [8, 6, 7, 5, 8, 7, 6, 9]
})

X = df.copy()

# Calculate VIF manually
for feature in X.columns:

    # Target feature
    y = X[feature]

    # Remaining input features
    other_features = X.drop(columns=[feature])

    # Train a regression model
    model = LinearRegression()
    model.fit(other_features, y)

    # Calculate R-squared
    r2 = model.score(other_features, y)

    # Calculate VIF
    if r2 >= 1:
        vif = float("inf")
    else:
        vif = 1 / (1 - r2)

    print(f"{feature}: VIF = {vif:.2f}")
```
